```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"Titanic-Dataset.csv")
```

```python
df.groupby("Pclass")[["Parch","SibSp",'Survived']].sum()
```

```python
df.groupby("Pclass")[["Parch","SibSp",'Survived']].sum().plot(kind='bar', figsize=(4,3))
plt.show()
```

```python
df.groupby("Pclass")[["Parch","SibSp",'Survived']].sum().plot(kind='bar', figsize=(4,8),subplots=True)
plt.show()
```

```python
df.groupby("Pclass")[["Parch","SibSp",'Survived']].sum().plot(kind='line', figsize=(4,3))
plt.show()
```

```python
details=df.groupby("Sex")[["Parch","SibSp",'Survived']].sum()
details
```

```python
details.index
```

```python
details.plot(y=["Parch","SibSp","Survived"],kind='bar', figsize=(10,4))
plt.show()
```

```python
df.groupby("Pclass")[["Parch","SibSp",'Survived']].sum().plot(kind='pie', figsize=(10,8),subplots=True,legend=True)
plt.show()
```

```python
(pd.crosstab(df["Pclass"],df["Sex"],normalize=True)*100).plot(kind='line',ylabel="Percentage",figsize=(4,3))
plt.title("Percentage bar chart")
```

# **Subplot**

```python
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.hist(df["Fare"],bins=20)

plt.subplot(2,2,2)
plt.hist(df["Age"],bins=20)

plt.subplot(2,2,3)
plt.scatter(df["Age"],df["Fare"],c="Red")

plt.subplot(2,2,4)
plt.bar(df.groupby("Survived")["Fare"].count().index,df.groupby("Survived")["Fare"].count().values,width=0.2,color=["Red","Green"])
plt.show()
```

```python
plt.figure(figsize=(10,2))
plt.subplot(1,2,1)
plt.hist(df["Fare"],bins=20)

plt.subplot(1,2,2)
plt.hist(df["Age"],bins=20)
plt.show()
```

```python
plt.figure(figsize=(10,3))
plt.subplot(1,3,1)
plt.hist(df["Age"],bins=20)

plt.subplot(1,3,2)
plt.scatter(df["Age"],df["Fare"],c="Red")

plt.subplot(1,3,3)
plt.bar(df.groupby("Survived")["Fare"].count().index,df.groupby("Survived")["Fare"].count().values,width=0.2,color=["Red","Green"])
```

```python
plt.figure(figsize=(10,3))
plt.subplot(1,4,1)
plt.hist(df["Fare"],bins=20)

plt.subplot(1,4,2)
plt.hist(df["Age"],bins=20)

plt.subplot(1,4,3)
plt.scatter(df["Age"],df["Fare"],c="Red")

plt.subplot(1,4,4)
plt.bar(df.groupby("Survived")["Fare"].count().index,df.groupby("Survived")["Fare"].count().values,width=0.2,color=["Red","Green"])
plt.show()
```

```python
fig , axes= plt.subplots(2,2,figsize=(10,5))

axes[0,0].hist(df["Fare"],bins=20)
axes[0,0].set_title("Fare")

axes[0,1].hist(df["Age"],bins=20)
axes[0,1].set_title("Age")

axes[1,0].scatter(df["Age"],df["Fare"],c="Red")


axes[1,1].bar(df.groupby("Survived")["Fare"].count().index,df.groupby("Survived")["Fare"].count().values,width=0.2,color=["Red","Green"])
plt.show()
```

```python
fig, axes= plt.subplots(1,4,figsize=(10,3))
axes[0].hist(df["Fare"],bins=20)
axes[1].hist(df["Age"],bins=20)
axes[2].scatter(df["Age"],df["Fare"],c="Red")
axes[3].bar(df.groupby("Survived")["Fare"].count().index,df.groupby("Survived")["Fare"].count().values,width=0.2,color=["Red","Green"])

plt.show()
```

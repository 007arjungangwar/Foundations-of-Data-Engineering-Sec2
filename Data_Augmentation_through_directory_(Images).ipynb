# **AUGMENT EACH CLASS SUBDIRECTORY SEPARATELY**

````python
import os
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img
````

````python
# PATHS
directory_path = "/content/My dataset"       # original dataset root (Cat, Dog, Horse subfolders)
augmented_path = "/content/Augmented Data"   # output root folder
````

````python
# count all files recursively (through every subdirectory)
def get_file_count(path):
    if not os.path.exists(path):
        return 0
    return sum(len(files) for _, _, files in os.walk(path))
````

````python
# count files per subdirectory / class

def get_file_count_per_class(path):
    counts = {}
    if not os.path.exists(path):
        return counts
    for class_name in sorted(os.listdir(path)):
        class_dir = os.path.join(path, class_name)
        if os.path.isdir(class_dir):
            counts[class_name] = len([
                f for f in os.listdir(class_dir)
                if os.path.isfile(os.path.join(class_dir, f))
            ])
    return counts
````

````python
# 1. COUNT IMAGES BEFORE AUGMENTATION

before_count = get_file_count(directory_path)
before_counts_per_class = get_file_count_per_class(directory_path)

print(f"Number of images before augmentation: {before_count}")
for class_name, count in before_counts_per_class.items():
    print(f"  {class_name}: {count} images")
````

````python
# 2. CONSTRUCT THE ImageDataGenerator
datagen = ImageDataGenerator(
    rotation_range=45,        # Random rotation between 0 and 45
    width_shift_range=0.2,    # % shift
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='reflect'       # nearest, constant, reflect, wrap
)
````

````python
# 3. AUGMENT EACH CLASS SUBDIRECTORY SEPARATELY
#    -> output goes to augmented_path/<class_name>/  e.g. Augmented Data/Cat,
#       Augmented Data/Dog, Augmented Data/Horse
#    -> no manual matching needed, folder names mirror the originals

num_augmented_per_image = 5   # how many augmented copies to make per original image

os.makedirs(augmented_path, exist_ok=True)

for class_name in sorted(os.listdir(directory_path)):
    class_dir = os.path.join(directory_path, class_name)
    if not os.path.isdir(class_dir):
        continue

    # Create the matching output folder, e.g. "Augmented Data/Dog"
    output_class_dir = os.path.join(augmented_path, class_name)
    os.makedirs(output_class_dir, exist_ok=True)

    image_files = [
        f for f in os.listdir(class_dir)
        if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

    # print(f"\nAugmenting class '{class_name}' ({len(image_files)} images)...")

    for img_name in image_files:
        img_path = os.path.join(class_dir, img_name)
        img = load_img(img_path, target_size=(256, 256))
        x = img_to_array(img)
        x = x.reshape((1,) + x.shape)   # generator expects a batch dimension

        i = 0
        for batch in datagen.flow(
            x,
            batch_size=1,
            save_to_dir=output_class_dir,
            save_prefix=f'aug_{class_name}',
            save_format='png'
        ):
            i += 1
            if i >= num_augmented_per_image:
                break
````

# **Simplified version of the code**

````python
import os
from tensorflow.keras.preprocessing.image import load_img, img_to_array

num_augmented = 5

os.makedirs(augmented_path, exist_ok=True)

for class_name in os.listdir(directory_path):

    class_path = os.path.join(directory_path, class_name)
    output_path = os.path.join(augmented_path, class_name)

    if not os.path.isdir(class_path):
        continue

    os.makedirs(output_path, exist_ok=True)

    for image_name in os.listdir(class_path):

        if image_name.lower().endswith((".jpg", ".jpeg", ".png")):

            image_path = os.path.join(class_path, image_name)

            image = load_img(image_path, target_size=(256, 256))
            image = img_to_array(image)
            image = image.reshape((1,) + image.shape)

            generator = datagen.flow(
                image,
                batch_size=1,
                save_to_dir=output_path,
                save_prefix="aug",
                save_format="png"
            )

            for i in range(num_augmented):
                next(generator)
````

````python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("/content/hand-x-ray.jpg")

plt.imshow(img)
````

Rotate

````python
rotated = img.rotate(30)

plt.imshow(rotated)
plt.axis("off");
````

Flip

````python
flipped = img.transpose(Image.FLIP_LEFT_RIGHT)

plt.imshow(flipped)
plt.axis("off");
````

Scale

````python
scaled = img.resize((300, 300))

plt.imshow(scaled)
plt.axis("off");
````

Crop

````python
cropped = img.crop((0, 0, 300, 300))

plt.imshow(cropped)
plt.axis("off");
````

````python
cropped = img.crop((100, 0, 300, 300))

plt.imshow(cropped)
plt.axis("off");
````

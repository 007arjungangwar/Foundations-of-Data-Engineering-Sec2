# Fetch 15-Year Daily Weather Data for Chennai

image.png

```python
from datetime import datetime
import pandas as pd
import requests

# Chennai / Minambakkam Weather Station ID
STATION_ID = "43279"
API_KEY = "0eb7097e7fmsh51f08c47a5e06e1p1fcb3djsnb6eb69eac7da"
HOST = "meteostat.p.rapidapi.com"

# API Endpoint for Daily Weather Data
url = "https://meteostat.p.rapidapi.com/stations/daily"

headers = {
    "x-rapidapi-key": API_KEY,
    "x-rapidapi-host": HOST,
    "Content-Type": "application/json",
}

all_years_data = []

# Loop from 2011 through 2025 (15 complete years)
for year in range(2011, 2026):
    print(f"Fetching weather data for year {year}...")

    querystring = {
        "station": STATION_ID,
        "start": f"{year}-01-01",
        "end": f"{year}-12-31",
        "tz": "Asia/Kolkata",
    }

    response = requests.get(url, headers=headers, params=querystring)

    if response.status_code == 200:
        json_resp = response.json()
        if "data" in json_resp:
            all_years_data.extend(json_resp["data"])
    else:
        print(f"Error fetching year {year}: {response.status_code}")
```

# Python script to fetch, clean, and combine batting stats for multiple Indian players into a single dataset

```python
import time
import pandas as pd
import requests

# 1. Define Indian Players and their Cricbuzz Player IDs
indian_players = {
    "KL Rahul": "8733",
    "Virat Kohli": "2538",
    "Rohit Sharma": "3472",
    "Rishabh Pant": "9308",
    "Shubman Gill": "11939",
    "Hardik Pandya": "9647",
    "Ravindra Jadeja": "587",
    "Suryakumar Yadav": "7915",
    "Jasprit Bumrah": "9311",
    "Shreyas Iyer": "9428",
}

headers = {
    "x-rapidapi-key": "0eb7097e7fmsh51f08c47a5e06e1p1fcb3djsnb6eb69eac7da",
    "x-rapidapi-host": "cricbuzz-cricket.p.rapidapi.com",
    "Content-Type": "application/json",
}

all_players_data = []

# 2. Iterate through each player
for name, player_id in indian_players.items():
    print(f"Fetching stats for: {name} (ID: {player_id})...")

    url = f"https://cricbuzz-cricket.p.rapidapi.com/stats/v1/player/{player_id}/batting"

    try:
        response = requests.get(url, headers=headers)

        if response.status_code == 200:
            data = response.json()

            if "values" in data and "headers" in data:
                # Extract rows & columns using the key-based approach
                rows = [item["values"] for item in data["values"]]
                cols = data["headers"]

                # Create DataFrame for this player
                df_player = pd.DataFrame(rows, columns=cols)
                df_player.rename(columns={"ROWHEADER": "Stat"}, inplace=True)

                # Add Player Identifier Columns
                df_player.insert(0, "Player_ID", player_id)
                df_player.insert(0, "Player_Name", name)

                all_players_data.append(df_player)
            else:
                print(f"No stats data found for {name}")

        else:
            print(f"Failed to fetch {name}: HTTP {response.status_code}")

    except Exception as e:
        print(f"Error fetching {name}: {e}")

    # Add a short delay (0.5s) to stay within API rate limits
    time.sleep(0.5)
```

```python
all_players_data
```

```python
# 3. Combine all individual DataFrames into a single Master DataFrame
if all_players_data:
    master_df = pd.concat(all_players_data, ignore_index=True)

    print("\n--- Master Dataset Summary ---")
    print("Total Rows:", master_df.shape[0])
    print("Total Columns:", master_df.shape[1])
    print("\nPreview:")
    print(master_df.head(12))

    # 4. Save to CSV file
    master_df.to_csv("all_indian_players_batting_stats.csv", index=False)
    print("Dataset successfully saved to 'all_indian_players_batting_stats.csv'!")
else:
    print("No data collected.")
```
